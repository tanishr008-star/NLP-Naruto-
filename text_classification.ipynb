```python
import pandas as pd
```

```python
df = pd.read_json("../data/jutsu.jsonl", lines = True)
```

```python
def filter_jutsu(jutsu):
    jutsu_list = jutsu.split(',')
    for jutsus in jutsu_list:
        if jutsus.endswith("jutsu"): 
            return (jutsus.strip())
```

```python
df['jutsu_type_simplified'] = df['jutsu_type'].apply(filter_jutsu)

df[df.isnull().any(axis=1)]

```
